```python
import re
import pandas as pd
```

```python
f = open('WhatsApp Chat with Babubhai family🙋🏼.txt', 'r', encoding='utf-8')
```

```python
data = f.read()
```

```python
print(data)
```

```python
pattern = r'\d{1,2}/\d{1,2}/\d{2,4},\s\d{1,2}:\d{2}\s(?:AM|PM)\s-\s'
```

```python
messages = re.split(pattern, data)[1:]
print(messages)
```

```python
dates = re.findall(pattern, data)
dates
```
